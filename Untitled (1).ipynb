<h1 style="text-align:left;">📖 Introduction</h1>

Agriculture is an important sector that plays a major role in the economy and provides food and raw materials for different industries. Crop production depends on various factors such as crop type, area cultivated, season, location, and production quantity.

This project focuses on analyzing agriculture crop production data to understand production patterns, compare different crops and regions, and identify useful trends from the available data. Data analysis techniques are used to convert the raw dataset into meaningful information through data cleaning, manipulation, analysis, and visualization.

<h1 style="text-align:left;">Problem Statement</h1>


Agricultural datasets contain a large amount of information about crops, production, cultivated area, and different regions. It can be difficult to understand production patterns and compare crops by examining the raw data directly.
Therefore, this project aims to analyze and manipulate agricultural crop production data to identify important trends, compare crop production, and generate meaningful insights that can help in understanding agricultural production patterns.

<h1 style="text-align:left;">🎯Project Objectives</h1>

Q1. Which state has the highest crop production?

```python
df.groupby("State")["Production_Tonnes"].sum().sort_values(ascending=False)
```

Q2. Which crop has the highest production?

```python
df.groupby("Crop")["Production_Tonnes"].sum().sort_values(ascending=False)
```

```python
df.groupby("Crop")["Production_Tonnes"].sum().sort_values(ascending=False).head()
```

Q3. Which year has recorded the highest production?

```python
df.groupby("Year")["Production_Tonnes"].sum().sort_values(ascending=False)
```

Q4 which season produce maximum crops

```python
df.groupby("Season")["Production_Tonnes"].sum().sort_values(ascending=False)
```

Q5 what is the average rainfall as per states?

```python
df.groupby("State")["Rainfall_mm"].mean()
```

Q6.which Irrigation gives maximum production?

```python
df.groupby("Irrigation_Type")["Production_Tonnes"].max()
```

```python
top_state = df.groupby("State")["Production_Tonnes"].sum().sort_values(ascending=False).head(10)
state=top_state.index
tones=top_state.values 

plt.figure(figsize=(12,6),facecolor=("lightyellow"))

top_state.plot(
    kind="bar",
    color=["red","green","blue","orange","purple","brown","pink","cyan","gold","teal"]
)

plt.title("Top 10 States by Crop Production")
plt.xlabel("State")
plt.ylabel("Production")

plt.xticks(rotation=25)

plt.show()
```

```python
top_crop = df.groupby("Crop")["Production_Tonnes"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6),facecolor=("yellow"))
top_crop.plot(
    kind="bar",
    color=["navy","orange","green","red","purple","gold","brown","pink","gray","olive"]
)

plt.title("Top 10 Crop Production")

plt.show()
```

```python
year = df.groupby("Year")["Production_Tonnes"].sum()

fig, ax = plt.subplots(figsize=(10,6))

fig.patch.set_facecolor("beige")
ax.set_facecolor("aliceblue")

ax.plot(year.index,
        year.values,
        color="red",
        marker="o",
        linewidth=3)

plt.title("Year-wise Crop Production")
plt.xlabel("Year")
plt.ylabel("Production")

plt.grid(True)

plt.show()
```

```python
season=df.groupby("Season")["Production_Tonnes"].sum()
plt.figure(figsize=(8,8))

colors=["gold","green","red","skyblue"]

plt.pie(season.values,
        labels=season.index,
        autopct="%1.1f%%",
        colors=colors)

plt.title("Production by Season")

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(df["Rainfall_mm"],
            df["Production_Tonnes"],
            color="purple")

plt.title("Rainfall vs Production")

plt.xlabel("Rainfall")

plt.ylabel("Production")

plt.show()
```

```python
profit=df.groupby("Crop")["Profit"].sum().abs().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,6))

plt.bar(profit.index,
        profit.values,
        color="crimson")

plt.xticks(rotation=45)

plt.title("Top Profitable Crops")

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10,6))

fig.patch.set_facecolor("lightcyan")
ax.set_facecolor("whitesmoke")

ax.scatter(df["Rainfall_mm"],
           df["Production_Tonnes"],
           color="purple")

plt.title("Rainfall vs Production")
plt.xlabel("Rainfall (mm)")
plt.ylabel("Production (Tonnes)")

plt.grid(True)

plt.show()
```

```python
df.groupby("Area_Hectares")["Profit"].sum().sort_values(ascending=False)
```

```python
year_revenue = df.groupby("Year")["Revenue"].sum()

plt.figure(figsize=(10,5), facecolor="lightblue")

plt.plot(year_revenue.index,
         year_revenue.values,
         color="blue",
         marker="o",
         linewidth=2)

plt.title("Year-wise Agricultural Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")

plt.grid(True)

plt.show()
```
